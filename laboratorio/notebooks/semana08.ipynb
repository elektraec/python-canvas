## Ejercicio 1. Monitoreo de diez mediciones

Registrar 10 mediciones. Calcular promedio, contar cuántas superan un límite y mostrar el mayor valor. Resolver con funciones, for y condicionales, todavía sin listas.

### Pistas de razonamiento
- Crear una función para decidir si una medición supera el límite.
- Inicializar suma, contador_alertas y máximo.
- Leer 10 valores con for.
- Actualizar suma, alertas y máximo.
- Calcular promedio y mostrar resultados.


```python
datos = [42, 51, 48, 60, 55, 39, 70, 45, 52, 49]
limite = 50
alertas = 0
for valor in datos:
    if valor > limite:
        alertas += 1
print("Alertas:", alertas)
```

**Resultado esperado:** El resultado depende de las 10 mediciones; comprobar manualmente con un conjunto pequeño antes de usarlo en clase.


## Ejercicio 2. Trazado y corrección

Analizar el siguiente código, identificar el error y corregirlo: contador debe aumentar hasta 5, pero el programa no termina.

### Pistas de razonamiento
- La condición depende de contador.
- Dentro del ciclo contador nunca cambia.
- Por eso contador <= 5 permanece verdadero.
- Agregar contador += 1 dentro del while.


```python
contador = 1
while contador <= 5:
    print(contador)
    contador += 1
```

**Resultado esperado:** La actualización del contador evita el ciclo infinito.
