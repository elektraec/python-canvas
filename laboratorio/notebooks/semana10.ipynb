## Ejercicio 1. Procesar una lista de mediciones

Dada la lista [12.5, 10.8, 15.2, 9.7, 13.1], calcular promedio, máximo y mínimo.

### Pistas de razonamiento
- La lista almacena varias mediciones en una sola variable.
- Usar sum() y len() para el promedio.
- Usar max() y min() para extremos.


```python
mediciones = [12.5, 10.8, 15.2, 9.7, 13.1]
print(sum(mediciones) / len(mediciones))
print(max(mediciones))
print(min(mediciones))
```

**Resultado esperado:** Promedio = 12.26; máximo = 15.2; mínimo = 9.7.


## Ejercicio 2. Construir una lista y contar alertas

Solicitar 5 datos, almacenarlos en una lista y contar cuántos son mayores que 50.

### Pistas de razonamiento
- Crear lista vacía.
- Usar append() dentro de for.
- Recorrer la lista para contar valores > 50.


```python
datos = [40, 55, 60, 20, 80]
alertas = sum(1 for x in datos if x > 50)
print(alertas)
```

**Resultado esperado:** Si se ingresan 40, 55, 60, 20, 80 → 3 valores mayores que 50.
