## Ejercicio 1. Matriz de mediciones

Representar mediciones de 3 equipos durante 4 periodos y calcular el promedio de cada equipo.

### Pistas de razonamiento
- Cada fila representa un equipo.
- Cada columna representa un periodo.
- Recorrer fila por fila y calcular sum(fila)/len(fila).


```python
mediciones = [[10,12,11,13],[20,18,22,21],[7,8,9,8]]
for i, fila in enumerate(mediciones):
    print(i+1, sum(fila)/len(fila))
```

**Resultado esperado:** Equipo 1 = 11.5; Equipo 2 = 20.25; Equipo 3 = 8.0.
