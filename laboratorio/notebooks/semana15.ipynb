## Ejercicio 2. Validar entrada con try/except

Solicitar un número real. Si el usuario ingresa texto no numérico, mostrar un mensaje y no detener abruptamente el programa.

### Pistas de razonamiento
- Encerrar la conversión float() en try.
- Capturar ValueError.
- Mostrar mensaje controlado.


```python
entrada = "abc"
try:
    print(float(entrada))
except ValueError:
    print("Error controlado")
```

**Resultado esperado:** Entrada 12.5 → se acepta; entrada abc → mensaje de error.
