## Ejercicio 1. Traducir el control de límite

Traducir a Python el algoritmo de la semana anterior que compara una medición con un límite.

### Pistas de razonamiento
- input() devuelve texto; convertir a float.
- Usar dos puntos al final de if y else.
- Indentar el bloque que pertenece a cada condición.


```python
medicion = 82
limite = 80
# Completa la condición
if medicion > limite:
    print("ALERTA")
else:
    print("NORMAL")
```

## Ejercicio 2. Costo total con validación básica

Leer cantidad y precio unitario. Calcular el total. Si el total es mayor o igual a 100, mostrar además «Compra de alto valor».

### Pistas de razonamiento
- Leer cantidad como int y precio como float.
- Calcular total = cantidad * precio.
- Mostrar total.
- Evaluar total >= 100.


```python
cantidad = 10
precio = 12.5
total = cantidad * precio
print("Total:", total)
if total >= 100:
    print("Compra de alto valor")
```

**Resultado esperado:** 10 unidades a 12.5 → total 125.0 y se muestra el mensaje adicional.
