## Ejercicio 1. Estadísticos descriptivos

Calcular media, mediana y desviación estándar de una lista de mediciones utilizando statistics.

### Pistas de razonamiento
- Importar statistics.
- Usar mean(), median() y stdev().
- Interpretar: media = centro; mediana = valor central; desviación = dispersión.


```python
import statistics
datos = [10, 12, 11, 15, 9, 13, 10]
print(statistics.mean(datos))
print(statistics.median(datos))
```

**Resultado esperado:** Media ≈ 11.43; mediana = 11; desviación estándar ≈ 2.07.
