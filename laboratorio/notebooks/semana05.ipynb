## Ejercicio 1. Validar una entrada

Solicitar una calificación entre 0 y 10. Mientras el usuario ingrese un valor fuera del rango, volver a solicitarlo.

### Pistas de razonamiento
- Leer el valor inicial.
- Mientras valor < 0 o valor > 10, mostrar error y volver a leer.
- Al terminar el ciclo, el dato es válido.


```python
nota = 8.5
while nota < 0 or nota > 10:
    print("Valor fuera de rango")
print("Nota valida:", nota)
```

**Resultado esperado:** Si se ingresan 15, -2 y 8.5, el programa rechaza los dos primeros y acepta 8.5.


## Ejercicio 2. Acumular hasta alcanzar una meta

Registrar valores positivos hasta que la suma acumulada alcance o supere 100. Mostrar cuántos valores fueron necesarios y la suma final.

### Pistas de razonamiento
- Inicializar suma = 0 y contador = 0.
- Mientras suma < 100, leer un valor.
- Acumular y aumentar contador.
- Al salir, mostrar resultados.


```python
valores = [30, 25, 20, 35]
suma = 0
contador = 0
i = 0
while suma < 100:
    valor = valores[i]
    i += 1
    if valor > 0:
        suma += valor
        contador += 1
print("Cantidad de valores:", contador)
print("Suma final:", suma)
```

**Resultado esperado:** Con 30, 25, 20 y 35 → 4 valores y suma 110.
