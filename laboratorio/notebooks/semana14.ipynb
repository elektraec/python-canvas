## Ejercicio 1. Búsqueda lineal

Buscar el valor 22 dentro de la lista [15, 18, 22, 13, 27] e indicar su posición.

### Pistas de razonamiento
- Recorrer la lista desde el inicio.
- Comparar cada elemento con el objetivo.
- Cuando coincida, guardar posición y detener la búsqueda.


```python
datos = [15, 18, 22, 13, 27]
print(datos.index(22))
```

## Ejercicio 2. Ordenar sin modificar el original

Dada una lista de mediciones, crear una nueva lista ordenada de menor a mayor sin alterar la original.

### Pistas de razonamiento
- sorted(lista) devuelve una nueva lista.
- sort() modifica la lista existente.
- Comparar ambas listas.


```python
original = [32, 18, 25, 40, 21]
ordenada = sorted(original)
print(ordenada)
```

**Resultado esperado:** Original sigue [32, 18, 25, 40, 21]; ordenada es [18, 21, 25, 32, 40].
