## Ejercicio 1. Función para calcular promedio

Crear una función que reciba tres valores y retorne su promedio. Luego usarla desde el programa principal.

### Pistas de razonamiento
- Definir la función promedio(a,b,c).
- Calcular y retornar el valor.
- Solicitar datos fuera de la función.
- Invocar y mostrar el resultado.


```python
def promedio(a, b, c):
    return (a + b + c) / 3

print(promedio(6, 8, 10))
```

**Resultado esperado:** Con 6, 8 y 10 → 8.0.


## Ejercicio 2. Programa modular de validación y cálculo

Crear una función es_valido(valor, minimo, maximo) que retorne True o False y otra función calcular_potencia(v, i).

### Pistas de razonamiento
- Separar la validación del cálculo.
- Cada función debe realizar una sola tarea.
- El programa principal decide qué hacer con el resultado.


```python
def calcular_potencia(voltaje, corriente):
    return voltaje * corriente

print(calcular_potencia(120, 2))
```

**Resultado esperado:** Con 120 V y 2 A → Potencia: 240.0.
