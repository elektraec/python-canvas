## Ejercicio 1. Leer un archivo de texto

Suponer que datos.txt contiene una medición por línea. Leer todas las líneas, convertirlas a float y calcular el promedio.

### Pistas de razonamiento
- Abrir con with open(...).
- Recorrer líneas.
- Aplicar strip() y float().
- Guardar en lista y calcular promedio.


```python
contenido = "10\n12\n14\n"
mediciones = [float(x) for x in contenido.splitlines()]
print(sum(mediciones)/len(mediciones))
```

**Resultado esperado:** Si el archivo contiene 10, 12 y 14 → promedio 12.0.


## Ejercicio 2. Procesar un CSV sencillo

Archivo mediciones.csv: codigo,valor. Mostrar el código y el valor de cada fila y calcular el promedio de valores.

### Pistas de razonamiento
- Importar csv.
- Usar csv.DictReader para trabajar con nombres de columnas.
- Convertir valor a float.
- Acumular en una lista.


```python
import csv, io
texto = "codigo,valor\nA,10\nB,12\nC,14\n"
filas = list(csv.DictReader(io.StringIO(texto)))
print(filas)
```

**Resultado esperado:** El resultado depende del archivo; el ejemplo incluido produce promedio 12.0.
