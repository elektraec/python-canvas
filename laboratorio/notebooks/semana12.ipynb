## Ejercicio 1. Normalizar un código

Leer un código ingresado por el usuario, eliminar espacios externos y convertirlo a mayúsculas.

### Pistas de razonamiento
- strip() elimina espacios iniciales y finales.
- upper() convierte a mayúsculas.
- Comparar el resultado con un formato esperado.


```python
codigo = " eq-15 "
print(codigo.strip().upper())
```

**Resultado esperado:** Entrada « eq-15 » → salida «EQ-15».


## Ejercicio 2. Buscar registros por prefijo

Dada una lista de códigos, mostrar únicamente los que empiezan por «TEMP».

### Pistas de razonamiento
- Recorrer la lista.
- Usar startswith('TEMP').
- Mostrar los registros que cumplen.


```python
codigos = ["TEMP-01", "MOT-02", "TEMP-03", "PRES-01"]
print([c for c in codigos if c.startswith("TEMP")])
```

**Resultado esperado:** TEMP-01 y TEMP-03.
