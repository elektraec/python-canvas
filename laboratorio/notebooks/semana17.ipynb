## Ejercicio 1. Mini sistema de procesamiento de mediciones

A partir de un archivo CSV con codigo,valor, leer los registros, calcular promedio, máximo y mínimo, contar valores que superan un límite y mostrar el código asociado al máximo.

### Pistas de razonamiento
- Crear función cargar_datos().
- Leer CSV y almacenar registros como diccionarios.
- Crear función resumen() para calcular indicadores.
- Recorrer registros para máximo y alertas.
- Probar con datos normales y un valor extremo.


```python
registros = [{"codigo":"S1","valor":42},{"codigo":"S2","valor":61},{"codigo":"S3","valor":55}]
print(max(registros, key=lambda r: r["valor"]))
```
